The model is evaluated against the traditional NLP baseline
and recurrent neural network models.

```python
import time
import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Create a simple PyTorch model for emotion classification
class EmotionClassifier(nn.Module):
    def __init__(self, vocab_size=10000, embedding_dim=128, num_classes=6):
        super(EmotionClassifier, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(embedding_dim, 64, batch_first=True, bidirectional=True)
        self.fc1 = nn.Linear(128, 64)
        self.fc2 = nn.Linear(64, num_classes)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.3)
        
    def forward(self, x):
        embedded = self.embedding(x)
        lstm_out, (h_n, c_n) = self.lstm(embedded)
        last_hidden = h_n[-1]  # Use last hidden state
        x = self.fc1(last_hidden)
        x = self.relu(x)
        x = self.dropout(x)
        x = self.fc2(x)
        return x

start = time.time()
model = EmotionClassifier().to(device)
print(f"Model created in {time.time() - start:.2f} seconds")
print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
```
